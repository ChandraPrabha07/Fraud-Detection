# Exploratory analysis — credit card fraud

Quick EDA of `data/raw/creditcard.csv` and a look at the saved training results. Run from the project root (`jupyter notebook notebooks/`).

```python
import sys, json
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))

import pandas as pd
from ml.preprocessing.loader import load_creditcard_data

df, info = load_creditcard_data(ROOT / 'data/raw/creditcard.csv')
info
```
